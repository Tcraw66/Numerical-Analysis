# NumPy Basics

NumPy is a Python library for fast numerical computing. Its main object is the **array**, which stores numbers in one or more dimensions.

```python
import numpy as np
```

## Creating arrays
Arrays can be made from Python lists or with helper functions such as `zeros` and `arange`.

```python
a = np.array([1, 2, 3, 4])
matrix = np.array([[1, 2], [3, 4]])
print(a)
print(matrix)
print("Matrix shape:", matrix.shape)
```

## Array operations
Math is applied element by element, without writing a loop.

```python
print(a + 10)
print(a * 2)
print("Mean:", a.mean())
print("Sum:", a.sum())
```

## Selecting values
Use indexes and conditions to select data.

```python
print("First value:", a[0])
print("Middle values:", a[1:3])
print("Values greater than 2:", a[a > 2])
```
