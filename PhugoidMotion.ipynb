# Phugoid Motion


The term **phugoid** in aeronautics refers to a **motion pattern where an aircraft oscillates up and down** —nose-up and climb, then nose-down and descend— around an equilibrium trajectory. 

### Variables Key

| Symbol | Meaning |
|---|---|
| $z$ | Vertical coordinate / height variable |
| $z_t$ | Reference (trim) value of $z$ corresponding to the trim velocity |
| $\theta$ | Flight-path angle |
| $s$ | Distance measured along the flight path (arc length) |
| $R$ | Radius of curvature of the flight path |
| $C$ | Arbitrary constant of integration |
| $\dfrac{dz}{ds}$ | Rate of change of $z$ with respect to distance along the flight path |
| $\dfrac{d\theta}{ds}$ | Rate of change of flight-path angle along the flight path |
| $\dfrac{d\theta}{dz}$ | Rate of change of flight-path angle with respect to $z$ |

Let $ds$ represent a small arc-length of the trajectory. We can write $1/R = dθ/ds$ and $sinθ = −dz/ds$

## Final Lanchester Phugoid Equations

### 1. Phugoid Equation

$$
\boxed{
\frac{z}{z_t}-\cos\theta=\frac{2z}{R}
}
$$

### 2. Chain-Rule Relationship

$$
\boxed{
\frac{1}{R}
=
\frac{d\theta}{ds}
=
\frac{dz}{ds}\frac{d\theta}{dz}
=
-\sin\theta\frac{d\theta}{dz}
}
$$

### 3. Exact-Derivative Form

$$
\boxed{
\frac{d}{dz}
\left(
\sqrt{z}\cos\theta
\right)
=
\frac{\sqrt{z}}{2z_t}
}
$$

### 4. Integrated Flight-Path Equation

$$
\boxed{
\cos\theta
=
\frac{1}{3}\frac{z}{z_t}
+
C\sqrt{\frac{z_t}{z}}
}
$$

### 5. Radius-of-Curvature Equation

$$
\boxed{
\frac{z_t}{R}
=
\frac{1}{3}
-
\frac{C}{2}
\left(
\frac{z_t}{z}
\right)^{3/2}
}
$$


# Reasoning



# Lanchester's Phugoid Theory

The following derivation follows Lanchester's treatment of the phugoid in *Aerodonetics*. The coordinate $z$ is taken as positive downward.

The phugoid equation is

$$
\boxed{
\frac{z}{z_t}-\cos\theta=\frac{2z}{R}
}
$$

where $z_t$ is the value of $z$ corresponding to the trim velocity, $\theta$ is the flight-path angle, and $R$ is the radius of curvature of the flight path.

Let $ds$ represent a small arc length of the trajectory. We have

$$
\boxed{
\frac{1}{R}=\frac{d\theta}{ds}
}
$$

and

$$
\boxed{
\sin\theta=-\frac{dz}{ds}.
}
$$

Employing the chain rule of calculus,

$$
\boxed{
\frac{1}{R}
=
\frac{d\theta}{ds}
=
\frac{dz}{ds}\frac{d\theta}{dz}
=
-\sin\theta\frac{d\theta}{dz}.
}
$$

Starting with the phugoid equation,

$$
\frac{z}{z_t}-\cos\theta=\frac{2z}{R},
$$

multiply through by

$$
\frac{1}{2\sqrt{z}}
$$

to obtain

$$
\boxed{
\frac{\sqrt{z}}{2z_t}
-
\frac{\cos\theta}{2\sqrt{z}}
=
\frac{\sqrt{z}}{R}.
}
$$

Substituting

$$
\frac{1}{R}
=
-\sin\theta\frac{d\theta}{dz}
$$

into the right-hand side gives

$$
\frac{\sqrt{z}}{2z_t}
-
\frac{\cos\theta}{2\sqrt{z}}
=
-\sqrt{z}\sin\theta\frac{d\theta}{dz}.
$$

Bringing the cosine term to the right-hand side,

$$
\boxed{
\frac{\sqrt{z}}{2z_t}
=
\frac{\cos\theta}{2\sqrt{z}}
-
\sqrt{z}\sin\theta\frac{d\theta}{dz}.
}
$$

The right-hand side is an exact derivative. Therefore,

$$
\boxed{
\frac{d}{dz}
\left(
\sqrt{z}\cos\theta
\right)
=
\frac{\sqrt{z}}{2z_t}.
}
$$

Integrating both sides,

$$
\sqrt{z}\cos\theta
=
\frac{z^{3/2}}{3z_t}
+
C\sqrt{z_t},
$$

where $C$ is an arbitrary constant of integration.

Dividing through by $\sqrt{z}$ gives

$$
\boxed{
\cos\theta
=
\frac{1}{3}\frac{z}{z_t}
+
C\sqrt{\frac{z_t}{z}}.
}
$$

This is the equation for the phugoid flight path in terms of $z$ and $\theta$.

---

## Radius of Curvature

Taking the derivative of

$$
\cos\theta
=
\frac{1}{3}\frac{z}{z_t}
+
C\sqrt{\frac{z_t}{z}}
$$

with respect to $z$ gives

$$
-\sin\theta\frac{d\theta}{dz}
=
\frac{1}{3z_t}
-
\frac{C}{2}
\sqrt{\frac{z_t}{z^3}}.
$$

Using

$$
\frac{1}{R}
=
-\sin\theta\frac{d\theta}{dz},
$$

we obtain

$$
\frac{1}{R}
=
\frac{1}{3z_t}
-
\frac{C}{2}
\sqrt{\frac{z_t}{z^3}}.
$$

Multiplying through by $z_t$,

$$
\boxed{
\frac{z_t}{R}
=
\frac{1}{3}
-
\frac{C}{2}
\left(
\frac{z_t}{z}
\right)^{3/2}.
}
$$

---

## Final Lanchester Equations

The key equations from the derivation are therefore

$$
\boxed{
\frac{z}{z_t}-\cos\theta=\frac{2z}{R}
}
$$

$$

\frac{d\theta}{ds}

$$
